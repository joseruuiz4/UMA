## 5. Raíces de la Unidad

```python
#Vamos a buscar las raíces cuartas de la unidad.
n=4
#Creamos un array con los valores [1,2,3,...,n]
k=np.arange(n)
#Calculamos las raíces
theta=2*np.pi*k/n
raices=np.exp(1j*theta)
for i, z in enumerate(raices):
    print(f"z_{i} = {z:.4f}")
```

```python
#Vamos a buscar las raíces sextas de la unidad.
n=6
#Creamos un array con los valores [1,2,3,...,n]
k=np.arange(n)
#Calculamos las raíces
theta=2*np.pi*k/n
raices=np.exp(1j*theta)
#Obtenemos la parte real y la parte imaginaria
real=raices.real
imag=raices.imag

for i in range(n):
    print(f"z_{i} = {real[i]:6.3f}+{imag[i]:6.3f}i")
```

```python
# Procedemos a visualizar las raíces sextas de la unidad en el plano
fig , ax=plt.subplots()

 # Primero, vamos a representar el círculo unitario
theta=np.linspace(0,2*np.pi,100)
ax.plot(np.cos(theta), np.sin(theta),'k--' ,alpha =0.3)

# Calculamos las raíces sextas como antes
k=np.arange(n)
theta=2*np.pi*k/n
raices=np.exp(1j*theta)

ax.scatter( raices.real, raices.imag, s=100 , c= 'red')
ax.set_aspect('equal')
ax.grid(True)
ax.set_xlabel('Parte Real')
ax.set_ylabel('Parte Imaginaria')
plt.show()
```

## Gráficos 3D y Superficies

### Creando Mallas 2D con Meshgrid

Para evaluar funciones $f(x,y)$, necesitamos mallas de coordenadas:

Ahora **X** e **Y** son arrays 2D representando todos los pares $(x,y)$.

```python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)

# Verificar forma de las mallas
print('Forma de X e Y:', X.shape)
```

### Graficando Superficies 3D

Superficie: $z = x^2 + 2y^2$

```python
# Definir campo escalar
Z = X**2 + 2*Y**2

# Crear gráfico 3D
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('$z = x^2 + 2y^2$')
plt.title('Superficie cuadrática')
plt.show()
```

## Curvas de Nivel y Curvas Paramétricas

### Líneas de Contorno

Curvas de nivel: $x^2+y^2=9$ y $x^2+y^2=16$

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig, ax = plt.subplots(figsize=(8, 8))
ax.contour(X, Y, Z, levels=[9, 16], colors='black')
ax.set_aspect('equal')
ax.set_title('$x^2 + y^2 = c$')
ax.grid(True)
plt.show()
```

## Matemáticas Simbólicas con SymPy

**SymPy** proporciona capacidades de matemáticas simbólicas:

- Cálculos simbólicos exactos
- Derivadas e integrales
- Resolución de ecuaciones
- Operaciones matriciales
- Simplificación y manipulación

### Calculando Derivadas

$f(x) = e^{x/2} \cdot \sin^2(x/3)$

SymPy devuelve la expresión simbólica exacta para $f'(x)$:

```python
# Definir variable simbólica
x = sp.symbols('x')

# Definir función
f = sp.exp(x/2) * sp.sin(x/3)**2

# Calcular derivada
df = f.diff(x)
print('f\' =')
sp.pprint(df)
```

### Derivadas Parciales y Gradientes

Para funciones multivariables $f(x,y)$:

Puedes calcular derivadas de cualquier orden.

```python
x, y = sp.symbols('x y')
f = x**2 + y**2

# Derivadas parciales de primer orden
df_dx = f.diff(x)    # 2*x
df_dy = f.diff(y)    # 2*y

# Derivadas parciales de segundo orden
d2f_dx2 = f.diff(x, x)   # 2
d2f_dxdy = f.diff(x, y)  # 0

print('∂f/∂x =', df_dx)
print('∂f/∂y =', df_dy)
print('∂²f/∂x² =', d2f_dx2)
print('∂²f/∂x∂y =', d2f_dxdy)
```

### Encontrando Puntos Críticos

Para encontrar los puntos críticos de $f(x,y) = y x^2 e^{xy}$:

```python
x, y = sp.symbols('x y')
f = y * x**2 * sp.exp(x*y)

# Calcular derivadas parciales
df_dx = f.diff(x)
df_dy = f.diff(y)

# Definir ecuaciones
eq1 = sp.Eq(df_dx, 0)
eq2 = sp.Eq(df_dy, 0)

# Resolver sistema
puntos_criticos = sp.solve([eq1, eq2], [x, y])
print('Puntos críticos:', puntos_criticos)
```
